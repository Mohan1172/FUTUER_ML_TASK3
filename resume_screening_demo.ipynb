Run these cells from the project folder after installing `requirements.txt`. This notebook demonstrates the core ranking logic without the web UI.


```python
from pathlib import Path
from resume_screening import screen_resumes, extract_text_from_upload
```

```python
job_description = Path('sample_data/job_description.txt').read_text(encoding='utf-8')
resume_files = ['sample_data/resume_python_developer.txt', 'sample_data/resume_data_analyst.txt']
resumes = [{'filename': Path(p).name, 'text': Path(p).read_text(encoding='utf-8')} for p in resume_files]
skills = ['python', 'sql', 'git', 'rest api', 'scikit-learn', 'pandas', 'machine learning', 'docker', 'excel', 'power bi', 'statistics', 'data visualization']
results = screen_resumes(job_description, resumes, skills=skills, required_skills=['python', 'sql', 'git'], preferred_skills=['scikit-learn', 'docker'])
results[['rank', 'filename', 'overall_score', 'skill_match_pct', 'text_similarity_pct', 'matched_skills', 'missing_skills']]
```

## Interpretation
The ranking is based on keyword/skill coverage and TF-IDF text similarity. Review each resume manually; the score is not a prediction of job performance.
